## Шаг 1. Как собирался датасет

**Источник:** iTunes Search API (`entity=song`, `media=music`, страна US).
Это публичный API без ключа, отдаёт 30-секундные preview в `.m4a`.

**Схема сбора:**
1. Задано 10 жанров × 5 seed-артистов = 50 «якорей».
2. Для каждого артиста запрашиваем до 15 треков, фильтруем:
   - есть `previewUrl`;
   - `artistName` содержит фамилию seed-артиста (регулярка `\b<last>\b`);
   - нет дублей по `trackId` и по нормализованному названию.
3. Лимиты: **≤ 2 трека на артиста**, **≤ 8 треков на жанр** → целевой размер ≈ 80, фактически 74.
4. Preview скачиваются в `audio/<track_id>.m4a` с ретраями (3 попытки, проверка размера > 10 КБ).
5. Метаданные сохраняются в `tracks_metadata.csv` (title, artist, album, seed_genre, itunes_genre,
   release_date, year, full_duration_s, audio_path).

**Почему так:**
- preview хватает для CLAP/librosa — модель обучена на коротких окнах;
- seed-артисты задают «эталон» жанра, а `seed_genre` становится ground truth для валидации zero-shot;
- пауза 3 с между запросами — чтобы не ловить 403 от iTunes.

```python
# Задание 1
import os, re, time, json, warnings
import numpy as np, pandas as pd, requests
import matplotlib.pyplot as plt, seaborn as sns
from tqdm.auto import tqdm
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
np.random.seed(42)

AUDIO_DIR = "audio"
os.makedirs(AUDIO_DIR, exist_ok=True)

TRACKS_PER_ARTIST = 2
TRACKS_PER_GENRE = 8


SEEDS = {
    "rock":       ["Queen", "Nirvana", "Led Zeppelin", "Arctic Monkeys", "Foo Fighters"],
    "pop":        ["Dua Lipa", "Taylor Swift", "Ed Sheeran", "Billie Eilish", "The Weeknd"],
    "hip-hop":    ["Eminem", "Kendrick Lamar", "Drake", "Nas", "Travis Scott"],
    "electronic": ["Daft Punk", "Aphex Twin", "Deadmau5", "Calvin Harris", "Chemical Brothers"],
    "jazz":       ["Miles Davis", "John Coltrane", "Norah Jones", "Ella Fitzgerald", "Dave Brubeck"],
    "classical":  ["Beethoven", "Bach", "Debussy", "Chopin", "Mozart"],
    "metal":      ["Metallica", "Iron Maiden", "Black Sabbath", "Slayer", "Rammstein"],
    "reggae":     ["Bob Marley", "Peter Tosh", "Shaggy", "Toots and the Maytals", "Jimmy Cliff"],
    "country":    ["Johnny Cash", "Dolly Parton", "Luke Combs", "Willie Nelson", "Shania Twain"],
    "ambient":    ["Brian Eno", "Tycho", "Bonobo", "Nils Frahm", "Hammock"],
}
GENRES = list(SEEDS)
def itunes_search(term, limit=15, retries=3):
    for i in range(retries):
        try:
            r = requests.get("https://itunes.apple.com/search",
                             params=dict(term=term, media="music", entity="song", limit=limit, country="US"),
                             timeout=20)
            if r.status_code == 200:
                return r.json().get("results", [])
        except Exception as e:
            print("retry", term, e)
        time.sleep(5 * (i + 1))
    return []

def name_matches(seed, artist_name):
    last = seed.split()[-1]
    return re.search(rf"\b{re.escape(last)}\b", artist_name, flags=re.I) is not None

rows, seen_ids = [], set()
for genre, artists in SEEDS.items():
    n_genre = 0
    for artist in artists:
        if n_genre >= TRACKS_PER_GENRE:
            break
        res = itunes_search(artist)
        time.sleep(3)  # лимиты API
        seen_titles, n_art = set(), 0
        for r in res:
            if n_art >= TRACKS_PER_ARTIST or n_genre >= TRACKS_PER_GENRE:
                break
            if not r.get("previewUrl") or r["trackId"] in seen_ids:
                continue
            if not name_matches(artist, r.get("artistName", "")):
                continue
            key = re.sub(r"\(.*?\)|\[.*?\]|-.*", "", r["trackName"]).strip().lower()
            if key in seen_titles:
                continue
            seen_titles.add(key); seen_ids.add(r["trackId"])
            rows.append(dict(
                track_id=r["trackId"], title=r["trackName"], artist=r["artistName"],
                seed_artist=artist, album=r.get("collectionName"),
                seed_genre=genre, itunes_genre=r.get("primaryGenreName"),
                release_date=r.get("releaseDate"), full_duration_s=r.get("trackTimeMillis", 0) / 1000,
                preview_url=r["previewUrl"]))
            n_art += 1; n_genre += 1
    print(f"{genre:11s}: {n_genre} треков")

df = pd.DataFrame(rows)
df["year"] = pd.to_datetime(df["release_date"], errors="coerce").dt.year
print("Всего треков:", len(df))

def download(url, path):
    if os.path.exists(path) and os.path.getsize(path) > 0:
        return True
    for i in range(3):
        try:
            r = requests.get(url, timeout=30)
            if r.status_code == 200 and len(r.content) > 10_000:
                open(path, "wb").write(r.content)
                return True
        except Exception:
            pass
        time.sleep(2)
    return False

paths, ok = [], []
for _, r in tqdm(df.iterrows(), total=len(df)):
    p = f"{AUDIO_DIR}/{r.track_id}.m4a"
    ok.append(download(r.preview_url, p)); paths.append(p)
df["audio_path"] = paths
df = df[ok].reset_index(drop=True)
df.to_csv("tracks_metadata.csv", index=False)

print("Скачано:", len(df))
print(df.groupby("seed_genre").size().to_string())
print("макс. треков на артиста:", df.groupby("artist").size().max())
print("уникальных артистов:", df["artist"].nunique())
df[["title", "artist", "seed_genre", "itunes_genre", "year"]].head(10)
```

## Шаг 2. Какие признаки выбраны и почему

Признаки делятся на три слоя:

| Слой | Признаки | Как получаем | Зачем |
|---|---|---|---|
| Семантический | genre, mood, context, atmosphere | CLAP zero-shot | понимание «о чём» трек без ручной разметки |
| Акустический | tempo (BPM), energy (0–1) | librosa | плавные переходы в миксах, градус плейлиста |
| Метаданные | decade | год релиза | ностальгические подборки |
| Эмбеддинги | CLAP audio 512-d, similar_artists (kNN) | CLAP + kNN | похожесть треков, cold start |

**Почему именно эти группы:**
- **genre** — якорь вкуса и фильтр (нужен для cold start и проверки модели).
- **mood** — главный драйвер плейлистов; CLAP умеет ловить его без жанровой привязки,
  что позволяет смешивать жанры в одной подборке.
- **context** (workout/study/sleep/party…) — ситуативные рекомендации; жанр тут плохой предиктор,
  а CLAP-текст работает напрямую.
- **atmosphere** (warm/epic/ethereal…) — тонкая текстура, которой нет в жанре.
- **tempo/energy** — числовые признаки для сегментации и «градуса» плейлиста.
- **decade** — объяснимая ностальгия и возрастные предпочтения.
- **similar_artists / CLAP-эмбеддинг** — для kNN-рекомендаций и объяснимости.

**Текстовые промпты** заданы шаблонами (`This is {} music.`, `This music sounds {}.` и т.д.),
что даёт стабильные текстовые эмбеддинги для zero-shot без разметки.

## Шаг 3. Модели для извлечения признаков

**CLAP (`laion/clap-htsat-unfused`)** — контрастивная модель «аудио ↔ текст».
Используется два энкодера:
- **audio encoder** — 512-мерный эмбеддинг из 3 окон по 10 с (усреднение + L2-нормировка);
- **text encoder** — эмбеддинги промптов вида `This is rock music.`.

Zero-shot классификация: `softmax(scale · EMB_audio @ EMB_textᵀ)`, где `scale = logit_scale_a.exp()`
(аналог температуры CLIP). Для каждой группы признаков берём argmax → метка, второй по вероятности → `_2nd`.

**librosa** — акустические признаки:
- `beat_track` → tempo (BPM);
- `feature.rms` → средняя громкость;
- `onset_strength` → «ударность»;
- `spectral_centroid` → «яркость»;
- `zero_crossing_rate` → шумность/перкуссивность.

`energy` = среднее процентилей (rms, onset, centroid) — устойчиво к разным шкалам.

**Обработка аудио:** 48 кГц, моно, первые 30 с. Нарезка на окна по 10 с,
отбрасываем окна короче 1 с; если после этого пусто — берём весь клип.

**Версии:** transformers с `processor(audio=...)`; для совместимости есть fallback на устаревший `audios=`
(он бросает `ValueError` в новых версиях — поэтому ловим `(TypeError, ValueError)`).

```python
# Задание 3
import torch, librosa
from transformers import ClapModel, ClapProcessor

# --- страховка: df в памяти мог остаться пустым ---
df = pd.read_csv("tracks_metadata.csv")
print("df перезагружен:", df.shape)

device = "cuda" if torch.cuda.is_available() else "cpu"
MODEL_ID = "laion/clap-htsat-unfused"
model = ClapModel.from_pretrained(MODEL_ID).to(device).eval()
processor = ClapProcessor.from_pretrained(MODEL_ID)
SR = 48000
print("device:", device)

def to_tensor(x):
    if isinstance(x, torch.Tensor):
        return x
    for a in ("pooler_output", "audio_embeds", "text_embeds"):
        if getattr(x, a, None) is not None:
            return getattr(x, a)
    return x[0]

@torch.no_grad()
def embed_audio(chunks):
    try:
        inp = processor(audio=chunks, sampling_rate=SR, return_tensors="pt")
    except (TypeError, ValueError):
        inp = processor(audios=chunks, sampling_rate=SR, return_tensors="pt")
    inp = {k: v.to(device) for k, v in inp.items()}
    return to_tensor(model.get_audio_features(**inp)).float().cpu().numpy()

@torch.no_grad()
def embed_text(texts):
    inp = processor(text=texts, return_tensors="pt", padding=True)
    inp = {k: v.to(device) for k, v in inp.items()}
    return to_tensor(model.get_text_features(**inp)).float().cpu().numpy()

def l2(x):
    return x / (np.linalg.norm(x, axis=-1, keepdims=True) + 1e-9)

# --- проверка, что аудио на месте ---
print("треков в df:", len(df))
print("файлов существует:", int(df.audio_path.map(os.path.exists).sum()), "из", len(df))

CHUNK = 10 * SR
emb_list, lib_rows, failed, errs = [], [], [], []

for i, r in tqdm(df.iterrows(), total=len(df)):
    try:
        y, _ = librosa.load(r.audio_path, sr=SR, mono=True, duration=30)
        chunks = [y[s:s + CHUNK] for s in range(0, max(len(y) - CHUNK // 2, 1), CHUNK)]
        chunks = [c for c in chunks if len(c) > SR] or [y]
        e = l2(embed_audio(chunks)).mean(0)
        tempo, _ = librosa.beat.beat_track(y=y, sr=SR)
        lib_rows.append(dict(
            tempo=float(np.atleast_1d(tempo)[0]),
            rms=float(librosa.feature.rms(y=y).mean()),
            onset=float(librosa.onset.onset_strength(y=y, sr=SR).mean()),
            centroid=float(librosa.feature.spectral_centroid(y=y, sr=SR).mean()),
            zcr=float(librosa.feature.zero_crossing_rate(y).mean())))
        emb_list.append(e)
    except Exception as ex:
        print("fail", r.title, "|", type(ex).__name__, "|", ex)
        failed.append(i)
        errs.append((r.title, type(ex).__name__, str(ex)))

df = df.drop(index=failed).reset_index(drop=True)
print("failed:", len(failed), "ok:", len(emb_list))
if not emb_list:
    for t, e, m in errs[:5]:
        print("EXAMPLE ERROR:", t, "|", e, "|", m)
    raise RuntimeError("Все треки упали — смотрите EXAMPLE ERROR выше")

EMB = l2(np.stack(emb_list))
df = pd.concat([df, pd.DataFrame(lib_rows)], axis=1)
np.save("clap_embeddings.npy", EMB)
print("EMB shape:", EMB.shape)

# --- Zero-shot по 4 группам признаков ---
scale = float(model.logit_scale_a.exp().item()) if hasattr(model, "logit_scale_a") else 100.0
PROBS = {}

def softmax(z):
    z = z - z.max(1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(1, keepdims=True)

for group, cfg in FEATURES.items():
    names = list(cfg["labels"])
    prompts = [cfg["template"].format(p) for p in cfg["labels"].values()]
    T = l2(embed_text(prompts))
    P = softmax(scale * EMB @ T.T)
    PROBS[group] = pd.DataFrame(P, columns=names)
    order = np.argsort(-P, axis=1)
    df[group] = [names[o[0]] for o in order]
    df[group + "_2nd"] = [names[o[1]] for o in order]

# --- tempo / energy / decade ---
def pct(s): return s.rank(pct=True)
df["energy"] = (pct(df.rms) + pct(df.onset) + pct(df.centroid)) / 3
df["tempo_bucket"] = pd.cut(df.tempo, [0, 90, 120, 150, 400],
                            labels=["slow", "medium", "fast", "very fast"])
df["decade"] = (df.year // 10 * 10).astype("Int64").astype(str) + "s"

df.to_csv("tracks_enriched.csv", index=False)
print(df[["title", "artist", "seed_genre", "genre", "mood", "context", "atmosphere",
          "tempo", "energy", "decade"]].head(15))

# --- Валидация zero-shot жанра ---
from sklearn.metrics import accuracy_score, confusion_matrix
acc = accuracy_score(df.seed_genre, df.genre)
print(f"Zero-shot genre accuracy: {acc:.2%} (случайно ≈ {1/len(GENRES):.0%})")
cm = confusion_matrix(df.seed_genre, df.genre, labels=GENRES)
plt.figure(figsize=(7, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=GENRES, yticklabels=GENRES)
plt.xlabel("CLAP predicted"); plt.ylabel("seed genre"); plt.title("Zero-shot genre confusion"); plt.show()
```

```python
import matplotlib.pyplot as plt, seaborn as sns
import numpy as np

# 1) mood по жанрам — стек-бар
ct = pd.crosstab(df["seed_genre"], df["mood"], normalize="index")
ct.plot(kind="bar", stacked=True, figsize=(12, 5), colormap="tab20")
plt.title("Распределение mood по seed-жанрам (доли)")
plt.ylabel("доля"); plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout(); plt.show()

# 2) energy по жанрам
plt.figure(figsize=(11, 4))
sns.boxplot(data=df, x="seed_genre", y="energy")
plt.xticks(rotation=45, ha="right"); plt.title("energy (0–1) по жанрам"); plt.tight_layout(); plt.show()

# 3) tempo по жанрам
plt.figure(figsize=(11, 4))
sns.boxplot(data=df, x="seed_genre", y="tempo")
plt.xticks(rotation=45, ha="right"); plt.title("tempo (BPM) по жанрам"); plt.tight_layout(); plt.show()

# 4) tempo vs energy, цвет = жанр
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="tempo", y="energy", hue="seed_genre", s=70)
plt.title("tempo vs energy, цвет = seed_genre"); plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout(); plt.show()

# 5) t-SNE CLAP-эмбеддингов
from sklearn.manifold import TSNE
xy = TSNE(n_components=2, perplexity=min(15, len(EMB) - 1), random_state=42, init="pca").fit_transform(EMB)
plt.figure(figsize=(9, 7))
sns.scatterplot(x=xy[:, 0], y=xy[:, 1], hue=df["seed_genre"], s=70)
plt.title("t-SNE CLAP-эмбеддингов (цвет = seed_genre)")
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left"); plt.tight_layout(); plt.show()
```

## Шаг 4. Какие эксперименты проведены

1. **Zero-shot жанр vs `seed_genre`** — основная валидация CLAP.
   Метрика: accuracy, confusion matrix. Baseline — случайное угадывание ≈ 10%.
2. **Распределения признаков** — mood/context/atmosphere по жанрам (стек-бары),
   tempo и energy (гистограммы + boxplot по жанрам).
3. **Корреляция CLAP-жанра и seed-жанра** — насколько текстовый промпт «This is X music»
   совпадает с реальной категорией.
4. **Согласованность mood ↔ energy** — ожидаем, что `aggressive/energetic` имеют высокий energy,
   `calm/sad` — низкий (проверка через boxplot).
5. **kNN по CLAP-эмбеддингам** — топ-5 похожих треков, доля соседей того же `seed_genre`.
   Это и прототип рекомендаций, и проверка качества эмбеддингов.
6. **similar_artists** — kNN по среднему эмбеддингу артиста; сравнение с ручными seed-артистами.

## Шаг 5. Визуализации и анализ признаков

- **Confusion matrix zero-shot жанра** — где CLAP путается (обычно rock↔metal, pop↔electronic).
- **Стек-бар `mood` по `seed_genre`** — есть ли у жанров характерные настроения.
- **Boxplot `energy` по жанрам** — согласуется ли акустика с ожиданиями.
- **Гистограмма `tempo`** — распределение по `tempo_bucket` (slow/medium/fast/very fast).
- **Scatter `tempo` vs `energy`**, цвет = жанр — кластеры жанров.
- **t-SNE / UMAP CLAP-эмбеддингов**, цвет = `seed_genre` — визуальная проверка семантики.
- **Таблица примеров**: title, artist, seed_genre, genre, mood, context, atmosphere, tempo, energy.

```python
# Задание 5
# Модель похожести: precision@5 по жанру для разных представлений
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import StandardScaler

def precision_at_k(X, labels, artists, k=5, no_same_artist=False):
    S = cosine_similarity(X)
    np.fill_diagonal(S, -np.inf)
    if no_same_artist:
        S[artists[:, None] == artists[None, :]] = -np.inf
    idx = np.argsort(-S, axis=1)[:, :k]
    return np.mean(labels[idx] == labels[:, None])

labels, artists = df.seed_genre.values, df.artist.values
z = lambda a: StandardScaler().fit_transform(a)

mood_X = z(PROBS["mood"].values)
ctx_X = z(PROBS["context"].values)
atm_X = z(PROBS["atmosphere"].values)
num_X = z(df[["tempo", "energy", "centroid", "zcr"]].values)
era_X = z(df[["year"]].fillna(df.year.median()).values)
tags_X = np.hstack([mood_X, ctx_X, atm_X, num_X])

REPS = {
    "CLAP embedding": EMB,
    "All tags + numeric": tags_X,
    "mood": mood_X,
    "context": ctx_X,
    "atmosphere": atm_X,
    "tempo/energy/timbre": num_X,
    "era (year)": era_X,
    "random": np.random.randn(len(df), 16),
}
res = pd.DataFrame({
    name: {"P@5": precision_at_k(X, labels, artists, 5),
           "P@5 (other artists)": precision_at_k(X, labels, artists, 5, True)}
    for name, X in REPS.items()}).T.sort_values("P@5 (other artists)", ascending=False)
display(res.round(3))
res.plot.barh(figsize=(8, 5)); plt.xlabel("precision@5 по жанру")
plt.title("Какие представления лучше отражают сходство")
plt.gca().invert_yaxis(); plt.show()

# Визуализация пространства эмбеддингов: UMAP и t-SNE
import umap
from sklearn.manifold import TSNE

xy_umap = umap.UMAP(n_neighbors=10, min_dist=0.3, metric="cosine", random_state=42).fit_transform(EMB)
xy_tsne = TSNE(perplexity=min(15, len(df) - 1), metric="cosine", init="random", random_state=42).fit_transform(EMB)

fig, axes = plt.subplots(2, 2, figsize=(16, 13))
for row, (name, xy) in enumerate([("UMAP", xy_umap), ("t-SNE", xy_tsne)]):
    for ax, col in zip(axes[row], ["seed_genre", "mood"]):
        sns.scatterplot(x=xy[:, 0], y=xy[:, 1], hue=df[col], palette="tab20", s=70, ax=ax)
        ax.set_title(f"{name} — цвет: {col}")
        ax.legend(fontsize=7, loc="best")
plt.tight_layout(); plt.show()

# Примеры ближайших треков
S = cosine_similarity(EMB); np.fill_diagonal(S, -np.inf)

def neighbors(i, k=5, no_same_artist=True):
    s = S[i].copy()
    if no_same_artist:
        s[df.artist.values == df.artist.values[i]] = -np.inf
    return np.argsort(-s)[:k]

rng = np.random.RandomState(1)
queries = rng.choice(len(df), 6, replace=False)
for q in queries:
    r = df.iloc[q]
    print(f"\n▶ {r.title} — {r.artist} [{r.seed_genre} | {r.mood} | {r.context} | {r.tempo:.0f} BPM]")
    for j in neighbors(q):
        n = df.iloc[j]
        print(f"    {S[q, j]:.2f}  {n.title} — {n.artist} [{n.seed_genre} | {n.mood} | {n.context} | {n.tempo:.0f} BPM]")

# Прослушать запрос и ближайших соседей
from IPython.display import Audio, display, HTML
q = queries[0]
display(HTML(f"<b>Запрос:</b> {df.title[q]} — {df.artist[q]}"))
display(Audio(df.audio_path[q]))
for j in neighbors(q, 3):
    display(HTML(f"<b>Сосед:</b> {df.title[j]} — {df.artist[j]} ({df.seed_genre[j]})"))
    display(Audio(df.audio_path[j]))

# Признак «похожие артисты»: kNN по усреднённым эмбеддингам артиста
art = df.groupby("artist").apply(lambda g: EMB[g.index].mean(0))
A = l2(np.stack(art.values)); names = art.index.to_list()
SA = cosine_similarity(A); np.fill_diagonal(SA, -np.inf)
similar_artists = {n: [names[j] for j in np.argsort(-SA[i])[:3]] for i, n in enumerate(names)}
for n in list(similar_artists)[:12]:
    print(f"{n:28s} → {similar_artists[n]}")
df["similar_artists"] = df.artist.map(similar_artists)
df.to_csv("tracks_enriched.csv", index=False)

# Какие признаки наиболее полезны для различения жанров (теги без жанра и без эмбеддинга)
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

feat_df = pd.concat([PROBS["mood"].add_prefix("mood:"), PROBS["context"].add_prefix("ctx:"),
                     PROBS["atmosphere"].add_prefix("atm:"),
                     df[["tempo", "energy", "centroid", "zcr", "year"]].fillna(0)], axis=1)
rf = RandomForestClassifier(n_estimators=300, random_state=42)
cv = cross_val_score(rf, feat_df, df.seed_genre, cv=min(5, df.seed_genre.value_counts().min()))
print(f"Genre из тегов, CV accuracy: {cv.mean():.2%}")
rf.fit(feat_df, df.seed_genre)
imp = pd.Series(rf.feature_importances_, feat_df.columns).sort_values(ascending=False).head(15)
imp.iloc[::-1].plot.barh(figsize=(7, 6), color="#8172B2")
plt.title("Топ-15 важных признаков для различения жанров")
plt.tight_layout(); plt.show()

print(res.round(3).to_string())
```

## Шаг 6. Выводы для задач музыкальных рекомендаций

1. **CLAP работает как zero-shot теггер.** Даже без разметки он даёт accuracy заметно выше случайной,
   а confusion matrix показывает системные путаницы (рок↔метал, поп↔электроника) — их стоит
   учитывать при построении фильтров: например, не штрафовать треки за «соседний» жанр.
2. **Mood и context — сильнее жанра для плейлистов.** CLAP-промпты `This music sounds ...` и
   `This is music for ...` дают ситуативные рекомендации, которые плохо выводятся из жанра.
   Это ключ к кросс-жанровым подборкам («энергичное для тренировки» может включать и хип-хоп, и рок).
3. **Эмбеддинги CLAP + kNN — готовый прототип рекомендателя.** Доля соседей того же жанра
   и примеры similar_artists показывают, что 512-d пространство осмысленно; это основа
   для «похожих треков», «похожих артистов» и cold start (новый трек без метаданных).
4. **Числовые признаки (tempo, energy) — для управления «градусом».** Их удобно использовать
   как фильтры/сортировку внутри семантически близкой группы, чтобы плейлист плавно нарастал.
5. **Decade** — дешёвый и объяснимый сигнал для ностальгических подборок; комбинируется
   с mood/energy (например, «спокойное 90-х»).

## Шаг 7. Ограничения и обнаруженные проблемы

**Данные:**
- Только 30-секундные preview → CLAP видит не весь трек; интро/аутро искажают эмбеддинг.
- iTunes Search API нестабилен: rate limit (пауза 3 с), иногда 0 результатов, часть треков не та,
  если фамилия артиста встречается у других исполнителей (фильтр по фамилии помогает, но не идеален).
- Дисбаланс: 74 трека на 10 жанров, по некоторым артистам < 2 треков; выборка мала для статистики.

**Метод:**
- CLAP обучен на общей музыке; на узких поджанрах (ambient, classical) путается сильнее.
- Zero-shot метки ограничены словарём промптов; mood/atmosphere — субъективны,
  «правильной» разметки нет, поэтому accuracy по жанру — единственная количественная проверка.
- `tempo` из librosa на 30-секундных клипах бывает неустойчив (особенно для ambient/classical).
- `energy` — прокси из акустики, а не перцептивная энергия; на тихих мастер-треках даёт смещение.
- `logit_scale` CLAP не калиброван под наши 10 классов → вероятности не откалиброваны,
  argmax надёжнее, чем абсолютные значения.

**Технические:**
- `.m4a` требует ffmpeg/audioread; на чистом окружении Colab иногда нужен `pip install soundfile` и `apt install ffmpeg`.
- Новые версии transformers убрали `audios=` в пользу `audio=` — старый код падает с `ValueError`,
  нужен try/except.
- CLAP на CPU медленный (~1 мин на 74 трека с 3 окнами) — для больших датасетов нужен GPU.

**Что улучшить:**
- Полные треки вместо preview (нужны лицензии/стриминг API).
- Больше seed-артистов и жанров, балансировка по числу треков.
- Валидация mood/context на ручной разметке (небольшой набор).
- UMAP/t-SNE для анализа кластеров и отбора «плохих» эмбеддингов.
- Калибровка вероятностей (temperature scaling) и кэширование CLAP-эмбеддингов между запусками.
